# Seeded Weekly Comparison — DQN vs DQN No-LLM vs Buy & Hold

**Test period:** 2024-01-05 → 2026-02-02  |  Weekly rebalancing  |  10% stop-loss  |  5 seeds  |  $10,000 starting capital

Metrics: `test_results/seeded_weekly_full/seeded_weekly_async_by_ticker.csv`. Shared code: `seeded_weekly_report.py`. Preview prices/dates: `data/*_weekly_stock_data.csv`.


```python
import os
from pathlib import Path
if Path.cwd().name == "notebooks" and (Path.cwd().parent / "run.py").exists():
    os.chdir(Path.cwd().parent)

from seeded_weekly_report import (
    DEFAULT_ASYNC_CSV,
    bah_preview,
    comparison_table,
    load_async_csv,
    max_return_table,
    plot_panels,
    print_scorecard,
    split_methods,
)

_, summary, bah_pct, TICKERS = load_async_csv(DEFAULT_ASYNC_CSV)
dqn, nollm = split_methods(summary, TICKERS)
START_DATE = "2024-01-01"

bah_df = bah_preview(TICKERS, bah_pct, start_date=START_DATE)
win0, win1 = bah_df["start_date"].iloc[0], bah_df["end_date"].iloc[0]
print(f"Test window: {win0}  →  {win1}")
bah_df[["ticker", "start_date", "end_date", "start_price", "end_price", "ret"]].rename(
    columns={"ret": "bah_return_pct"}
)
```

```python
display(comparison_table(summary, bah_df, TICKERS))
print("\n🟦 DQN (LLM)   ⬜ No-LLM   🟨 Buy & Hold")
```

```python
import matplotlib.pyplot as plt

bah_ret = bah_df.set_index("ticker").loc[TICKERS, "ret"].astype(float).tolist()
plot_panels(dqn, nollm, bah_ret, TICKERS)
plt.show()
```

```python
bah_ret_series = bah_df.set_index("ticker")["ret"].astype(float)
print_scorecard(dqn, nollm, bah_ret_series, TICKERS)
```
